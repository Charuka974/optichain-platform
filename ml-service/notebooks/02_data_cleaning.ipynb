# Supply Chain Risk Prediction: Data Cleaning
This notebook focuses on removing features that introduce data leakage, contain PII, or lack predictive value.

```python
# --- Setup & Load Raw Data ---
import pandas as pd

df = pd.read_csv("../data/DataCoSupplyChainDataset.csv", encoding="latin-1")
print("Original shape:", df.shape)
```

## Define Columns to Drop
Identifying leakage columns, personal information, and raw IDs.

```python
# --- Define Columns to Drop ---
# Leakage, PII, and useless ID/Text columns
leakage_cols = [
    "Days for shipping (real)", 
    "Delivery Status", 
    "shipping date (DateOrders)"
]

pii_cols = [
    "Customer Email", "Customer Fname", "Customer Lname", 
    "Customer Password", "Customer Street"
]

id_and_text_cols = [
    "Customer Id", "Order Customer Id", "Order Id", "Order Item Id",
    "Order Item Cardprod Id", "Product Card Id", "Product Category Id",
    "Product Description", "Product Image", "Product Name",
    "Customer Zipcode", "Order Zipcode", "Latitude", "Longitude",
    "Category Id", "Department Id"
]

drop_cols = leakage_cols + pii_cols + id_and_text_cols
drop_cols = [c for c in drop_cols if c in df.columns]
```

## Apply Drops & Save
Removing the columns and saving the cleaned dataset for the next phase.

```python
# --- Apply Drops ---
df_cleaned = df.drop(columns=drop_cols)
print(f"Dropped {len(drop_cols)} columns. Remaining shape:", df_cleaned.shape)

# --- Save Cleaned Data ---
# Saving to the cleaned data directory
df_cleaned.to_csv("../data/cleaned/cleaned_data.csv", index=False)
print("Cleaned data saved to ../data/cleaned/cleaned_data.csv")
```
