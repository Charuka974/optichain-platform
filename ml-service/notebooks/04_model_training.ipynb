# Supply Chain Risk Prediction: Model Training
Initializing and training classification models on the engineered dataset.

```python
# --- Setup & Load Data ---
import joblib
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

X_train = joblib.load('../data/train/X_train.pkl')
y_train = joblib.load('../data/train/y_train.pkl')
```

## Initialize Models
Setting up models with balanced class weights.

```python
# --- Initialize Models ---
class_0_count = (y_train == 0).sum()
class_1_count = (y_train == 1).sum()
xgb_scale_weight = class_0_count / class_1_count

models = {
    "Logistic Regression": LogisticRegression(
        max_iter=1000, 
        class_weight="balanced"
    ),
    "Random Forest": RandomForestClassifier(
        n_estimators=200, 
        random_state=42, 
        n_jobs=-1, 
        class_weight="balanced"
    ),
    "XGBoost": XGBClassifier(
        n_estimators=200, 
        eval_metric="logloss", 
        random_state=42,
        scale_pos_weight=xgb_scale_weight
    )
}
```

## Train and Export Models
Fitting the algorithms and saving the output file for evaluation.

```python
# --- Train Models ---
trained_models = {}
for name, model in models.items():
    print(f"Training {name}...")
    model.fit(X_train, y_train)
    trained_models[name] = model

# --- Save Trained Models ---
joblib.dump(trained_models, '../model/all_trained_models.pkl')
print("Models trained and saved to ../model/all_trained_models.pkl")
```
